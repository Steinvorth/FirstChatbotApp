# Learn Langfuse with OpenAI

Learning how to use langfuse for traceability with a simple request.

## 2) Add environment variables

Set these in your root `.env` file:

- `OPENAI_API_KEY`
- `LANGFUSE_PUBLIC_KEY`
- `LANGFUSE_SECRET_KEY`
- `LANGFUSE_HOST` (for cloud: `https://cloud.langfuse.com`)

```python
import os
from dotenv import load_dotenv
from langfuse import get_client
from langfuse.openai import OpenAI

load_dotenv('../.env')

requiredKeys = [
    'OPENAI_API_KEY',
    'LANGFUSE_PUBLIC_KEY',
    'LANGFUSE_SECRET_KEY',
]
missingKeys = [key for key in requiredKeys if not os.getenv(key)]
if missingKeys:
    raise ValueError(f'Missing env vars: {missingKeys}')

print('Environment variables loaded.')
```

## 3) Figuring out how to send a traced request based on a simple user and system prompt.

```python
openaiClient = OpenAI()
langfuseClient = get_client()

systemPrompt = 'You are a concise learning assistant. Explain clearly in 2 bullet points.'
userPrompt = 'What is Langfuse, and why is it useful when building AI apps?'

response = openaiClient.chat.completions.create(
    model='gpt-4o-mini',
    messages=[
        {'role': 'system', 'content': systemPrompt},
        {'role': 'user', 'content': userPrompt},
    ]
)

print(response.choices[0].message.content)

# Flush background events so the trace appears quickly in Langfuse.
langfuseClient.flush()
print('Trace sent. Open your Langfuse project to inspect it.')
```
